```python
import pandas as pd
```

```python
#Importing the dataset
df=pd.read_csv("wdbc.data")
```

```python
df
```

```python
#From the wdbc.names documentation we find out the columns names and assign them here
cols=['id','diagnosis','radius_mean','texture_mean','perimeter_mean','area_mean','smoothness_mean',
      'compactness_mean','concavity_mean','concave_points_mean','symmetry_mean','fractal_dimension_mean',
      'radius_se','texture_se','perimeter_se','area_se','smoothness_se',
      'compactness_se','concavity_se','concave_points_se','symmetry_se','fractal_dimension_se',
      'radius_worst','texture_worst','perimeter_worst','area_worst','smoothness_worst',
      'compactness_worst','concavity_worst','concave_points_worst','symmetry_worst','fractal_dimension_worst']
df.columns=cols
#Since we already have index and id is not a feature we can just drop it
df.drop("id",inplace=True,axis=1)
```

```python
#Data Preprocessing
```

```python
#Since the diagnosis feature is in the form of character we will convert it to numerical form so that later we can apply
#logistic regression to it
df["diagnosis"]=df["diagnosis"].map({"M":1,"B":0})
df
```

```python
#Data Visualization
import matplotlib.pyplot as plt
df.hist(bins=50,figsize=(12,15))
plt.show()
```

```python
#All the model have aprroximate bell-shaped graphs and some have slightly heavy tails so there is no point in doing log
#transformations
```

```python
#Since all the optimal features for the predictions are already present there is no point in adding new features
```

```python
#Training-Validation split
df1=df.copy()
df1=df.sample(frac=1,random_state=42)
train_size=int(0.8*len(df))
df_train=df1[:train_size].copy()
df_val=df1[train_size:].copy()
```

```python
#There is no need for imputation as there is no missing data
```

```python
#Encoding is also not required as there is only one categorical column which has been modified
```

```python
#Partioning the dataset into features and labels
y_train=df_train['diagnosis']
y_val=df_val["diagnosis"]
X_train=df_train.drop("diagnosis",axis=1)
X_val=df_val.drop("diagnosis",axis=1)
```

```python
#Standard Scaling
num_columns = [ col for col in df.columns if col != "diagnosis"]
train_mean=X_train[num_columns].mean()
train_std=X_train[num_columns].std()
X_train[num_columns]=(X_train[num_columns]-train_mean)/train_std
X_val[num_columns]=(X_val[num_columns]-train_mean)/train_std
```

```python
#Training using Scikit-Learn
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
import time
start=time.time()                                  #Time Start
model.fit(X_train,y_train)
end=time.time()                                    #Time Finish


#Validation Testing


threshold=0.45
val_probab=model.predict_proba(X_val)[:,1]
val_preds=(val_probab>=threshold)

from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,roc_auc_score,roc_curve
print(f"Accuracy:{accuracy_score(y_val,val_preds)}")
print(f"Precision:{precision_score(y_val,val_preds)}")
print(f"Recall:{recall_score(y_val,val_preds)}")
print(f"F1 Score:{f1_score(y_val,val_preds)}")
print(f"Confusion Matrix:{confusion_matrix(y_val,val_preds)}")
print(f"ROC_AUC:{roc_auc_score(y_val,val_probab)}")
fpr,tpr,thresholds=roc_curve(val_preds,y_val)
plt.plot(fpr,tpr,label="ROC Curve")
plt.plot([0,1],[0,1],"k:",label="Random Classifier's ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.title("ROC Curve")
plt.show()
print(f"Convergence Time:{end-start:.2f}seconds")
```
