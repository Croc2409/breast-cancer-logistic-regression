```python
#Training-Validation split
df1=df.copy()
df1=df.sample(frac=1,random_state=42)
train_size=int(0.8*len(df))
df_train=df1[:train_size].copy()
df_val=df1[train_size:].copy()
```

```python
#There is no need for imputation as there is no missing data
```

```python
#Encoding is also not required as there is only one categorical column which has been modified
```

```python
#Partioning the dataset into features and labels
y_train=df_train['diagnosis']
y_val=df_val["diagnosis"]
X_train=df_train.drop("diagnosis",axis=1)
X_val=df_val.drop("diagnosis",axis=1)
```

```python
#Standard Scaling
num_columns = [ col for col in df.columns if col != "diagnosis"]
train_mean=X_train[num_columns].mean()
train_std=X_train[num_columns].std()
X_train[num_columns]=(X_train[num_columns]-train_mean)/train_std
X_val[num_columns]=(X_val[num_columns]-train_mean)/train_std
```

```python
#Gradient Descent using Pure Python(Loss Function:Cross Entropy)
import time 
start=time.time()
weights=[0.0 for i in range(len(df_train.columns))]
bias=0.0
n_epochs=10000
learning_rate=0.01
losses=list()
import math
def cross_entropy(y_train,y_preds):
  sum=0
  a=1e-10
  for i in range(len(y_train)):
    sum+=(y_train[i]*(math.log(y_preds[i]+a))+((1-y_train[i])*(math.log(1-y_preds[i]+a))))
  return (-1)*sum/len(y_train)
def sigmoid(weighted_sum):
  return 1/(1+math.exp(-weighted_sum))

X1_train=X_train.copy()
X_train_list=X1_train.values.tolist()
y_train_list=y_train.values.tolist()
prev_loss=0

for epoch in range(n_epochs):
   y_preds=list()
   for i in range(len(X_train)):
    y_pred=0
    for j in range(len(X_train.columns)):                                             
      y_pred+=weights[j]*X_train_list[i][j]
    y_pred+=bias
    y_preds.append(sigmoid(y_pred))
   loss=cross_entropy(y_train_list,y_preds)
   losses.append(loss)
   bias_gradient=0.0
   for i in range(len(X_train)):
    bias_gradient+= y_preds[i]-y_train_list[i]
   bias_gradient=bias_gradient/len(X_train)
   gradient=list()
   for j in range(len(X_train.columns)):
    summation=0
    for i in range(len(X_train)):
      summation+=(y_preds[i]-y_train_list[i])*(X_train_list[i][j])
    gradient.append(summation/(len(X_train)))
   
   for i in range(len(X_train.columns)):
    weights[i]-=gradient[i]*learning_rate
   bias-=bias_gradient*learning_rate

   if epoch>0 and abs(prev_loss-loss)<1e-5:
    print(f"Converged at epoch:{epoch}")
    break

   print(f"Epoch:{epoch},Loss{loss}")
   prev_loss=loss
end=time.time()
```

```python
#Convergence Plot
plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Cross-Entropy Loss")
plt.title("Convergence Curve(Pure Python)")
plt.show()
```

```python
#Metrics defined in Pure Python

def accuracy(y_true,y_preds):
  correct=0
  for i in range(len(y_true)):
     if y_true[i]==y_preds[i]:
      correct+=1
  return correct/len(y_true)

def precision(y_true,y_preds):
  tp=0
  fp=0
  for i in range(len(y_true)):
    if y_preds[i]==1:
      if y_true[i]==1:
        tp+=1
      elif y_true[i]==0:
        fp+=1
  return tp/(tp+fp)

def recall(y_true,y_preds):
  tp=0
  fn=0
  for i in range(len(y_true)):
   if y_preds[i]==1 and y_true[i]==1:
    tp+=1
   elif y_preds[i]==0 and y_true[i]==1:
    fn+=1
  return tp/(tp+fn)

def f1_score(y_true,y_preds):
  p=precision(y_true,y_preds)
  r=recall(y_true,y_preds)
  return (2*p*r)/(p+r)

def confusion_matrix(y_true,y_preds):
  tp=0
  fn=0
  tn=0
  fp=0
  for i in range(len(y_true)):
    if y_true[i]==1:
      if y_preds[i]==1:
        tp+=1
      else:
        fn+=1
    else:
      if y_preds[i]==1:
        fp+=1
      else:
        tn+=1
  return [[tn,fp],
          [fn,tp]]

def ROC_AUC(y_probab,y_true):
  sorted_pairs=sorted(zip(y_probab,y_true),reverse=True)
  total_positives=sum(y_true)
  total_negatives=len(y_true)-total_positives
  tp=0
  fp=0
  tp_prev=0
  fp_prev=0
  auc=0
  for probab,true_value in sorted_pairs:
    if true_value==1:
      tp+=1
    else:
      fp+=1
    tpr=tp/total_positives
    tpr_prev=tp_prev/total_positives
    fpr=fp/total_negatives
    fpr_prev=fp_prev/total_negatives
    auc+=(fpr-fpr_prev)*(tpr+tpr_prev)/2
    fp_prev=fp
    tp_prev=tp
  return auc

def ROC_curve(y_probab,y_true):
  sorted_pairs=sorted(zip(y_probab,y_true),reverse=True)
  total_positives=sum(y_true)
  total_negatives=len(y_true)-total_positives
  tp=0
  fp=0
  tpr_list=list()
  fpr_list=list()
  for probab,true_value in sorted_pairs:
    if true_value==1:
      tp+=1
    else:
      fp+=1
    tpr_list.append(tp/total_positives)
    fpr_list.append(fp/total_negatives)
    
  plt.plot(fpr_list,tpr_list,label="ROC Curve")
  plt.plot([0,1],[0,1],"k:",label="Random Classifier's ROC Curve")
  plt.xlabel("False Positive Rate")
  plt.ylabel("True Positive Rate")
  plt.legend()
  plt.title("ROC Curve")
  plt.show()

           
```

```python
#Validation Testing

threshold=0.45
y_val_list=y_val.values.tolist()
X_val_list=X_val.values.tolist()
val_preds=list()
val_probab=list()
for i in range(len(X_val)):
    y_pred=0
    for j in range(len(X_val.columns)):
      y_pred+=weights[j]*X_val_list[i][j]
    y_pred+=bias
    val_probab.append(sigmoid(y_pred))
    if sigmoid(y_pred)>threshold:
      val_preds.append(1)
    else:
      val_preds.append(0)
print(f"Accuracy:{accuracy(y_val_list,val_preds)}")
print(f"Precision:{precision(y_val_list,val_preds)}")
print(f"Recall:{recall(y_val_list,val_preds)}")
print(f"F1 Score:{f1_score(y_val_list,val_preds)}")
print(f"Confusion Matrix:{confusion_matrix(y_val_list,val_preds)}")
print(f"ROC_AUC:{ROC_AUC(val_probab,y_val_list)}")
print(f"Convergence Time:{end-start:.2f}")
ROC_curve(val_probab,y_val_list)
```
