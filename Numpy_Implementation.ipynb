```python
#Training-Validation split
df1=df.copy()
df1=df.sample(frac=1,random_state=42)
train_size=int(0.8*len(df))
df_train=df1[:train_size].copy()
df_val=df1[train_size:].copy()
```

```python
#There is no need for imputation as there is no missing data
```

```python
#Encoding is also not required as there is only one categorical column which has been modified
```

```python
#Partioning the dataset into features and labels
y_train=df_train['diagnosis']
y_val=df_val["diagnosis"]
X_train=df_train.drop("diagnosis",axis=1)
X_val=df_val.drop("diagnosis",axis=1)
```

```python
#Standard Scaling
num_columns = [ col for col in df.columns if col != "diagnosis"]
train_mean=X_train[num_columns].mean()
train_std=X_train[num_columns].std()
X_train[num_columns]=(X_train[num_columns]-train_mean)/train_std
X_val[num_columns]=(X_val[num_columns]-train_mean)/train_std
```

```python
#Gradient Descent Using Numpy(Loss Function:Cross_Entropy)
import numpy as np
import time                                                                        #Time Start
start=time.time()
weights=np.zeros(len(X_train.columns))                                         
bias=0.0
n_epochs=10000
learning_rate=0.01
losses=list()
def cross_entropy(y_train,y_preds):
  a=1e-10
  return -np.mean(y_train*np.log(y_preds+a)+ (1-y_train)*(np.log(1-y_preds+a)))
def sigmoid(y_weighted_sum):
   return 1/(1+np.exp(-y_weighted_sum))
prev_loss=0
for epoch in range(n_epochs):
  y_preds=X_train@weights+bias                                                     #Calculating Loss
  y_preds=sigmoid(y_preds)
  loss=cross_entropy(y_train,y_preds)
  losses.append(loss)
  error=y_preds-y_train
  bias_gradient=np.mean(error)
  gradient=(1/len(X_train))*(X_train.T@error)
  weights-=gradient*learning_rate                                                  #Updating weights
  bias-=bias_gradient*learning_rate
  if epoch>0 and abs(prev_loss-loss)<1e-5:                                         #Convergence condition                                                        
    print(f"Converged at epoch:{epoch}")
    break
  print(f"Epoch:{epoch},Loss:{loss}")
  prev_loss=loss
end=time.time()                                                                    #Time end
```

```python
#Convergence Plot
plt.plot(losses)
plt.xlabel("Epochs")
plt.ylabel("Cross-Entropy Loss")
plt.title("Convergence Curve(Numpy)")
plt.show()
```

```python
#Metrics defined in numpy
def accuracy(y_true,y_preds):
  return np.mean(y_true==y_preds)
def precision(y_true,y_preds):
  tp=np.sum((y_true==1)&(y_preds==1))
  fp=np.sum((y_true==0)&(y_preds==1))
  return tp/(tp+fp)
def recall(y_true,y_preds):
  tp=np.sum((y_true==1)&(y_preds==1))
  fn=np.sum((y_true==1)&(y_preds==0))
  return tp/(tp+fn)
def f1_score(y_true,y_preds):
  p=precision(y_true,y_preds)
  r=recall(y_true,y_preds)
  return (2*p*r)/(p+r)
def confusion_matrix(y_true,y_preds):
  tp=np.sum((y_true==1)&(y_preds==1))
  fp=np.sum((y_true==0)&(y_preds==1))
  fn=np.sum((y_true==1)&(y_preds==0))
  tn=np.sum((y_true==0)&(y_preds==0))
  return [[tn.item(),fp.item()],
          [fn.item(),tp.item()]]
  
def ROC_AUC(y_probab,y_true):
  sorted_index=np.argsort(y_probab)[::-1]
  y_true_sorted=y_true[sorted_index]
  total_positive=np.sum(y_true)
  total_negative=len(y_true)-total_positive

  tp=0
  fp=0
  tpr_list=list()
  fpr_list=list()

  for true_value in y_true_sorted:
    if true_value==1:
      tp+=1
    else:
      fp+=1
    tpr_list.append(tp/total_positive)
    fpr_list.append(fp/total_negative)
  return np.trapezoid(tpr_list,fpr_list)

def ROC_curve(y_probab,y_true):
  sorted_index=np.argsort(y_probab)[::-1]
  y_true_sorted=y_true[sorted_index]
  total_positive=np.sum(y_true)
  total_negative=len(y_true)-total_positive

  tp=0
  fp=0
  tpr_list=list()
  fpr_list=list()

  for true_value in y_true_sorted:
    if true_value==1:
      tp+=1
    else:
      fp+=1
    tpr_list.append(tp/total_positive)
    fpr_list.append(fp/total_negative)
  
  plt.plot(fpr_list,tpr_list,label="ROC Curve")
  plt.plot([0,1],[0,1],"k:",label="Random Classifier's ROC Curve")
  plt.xlabel("False Positive Rate")
  plt.ylabel("True Positive Rate")
  plt.legend()
  plt.title("ROC Curve")
  plt.show()  
```

```python
#Validation Testing


threshold=0.45
val_probab=sigmoid(X_val@weights+bias)
val_preds=(val_probab>=threshold).astype(int)
   
print(f"Accuracy:{accuracy(y_val,val_preds)}")
print(f"Precision:{precision(y_val,val_preds)}")
print(f"Recall:{recall(y_val,val_preds)}")
print(f"F1 Score:{f1_score(y_val,val_preds)}")
print(f"Confusion Matrix:{confusion_matrix(y_val,val_preds)}")
print(f"ROC_AUC:{ROC_AUC(val_probab,y_val.values)}")
print(f"Convergence Time:{end-start:.2f}seconds")
ROC_curve(val_probab,y_val.values)
```
